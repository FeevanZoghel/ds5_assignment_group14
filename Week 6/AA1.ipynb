```python
def dropnan(kolom)->pd.DataFrame:
    '''
    verwijderd alle rijen met NaNwaarden uit de opgegeven kolom

    args: kolmo(str): Naam van de kolom


    Returns: pd.DataFrame zonder de Nan waardes

    '''
    return df.dropna(subset = [kolom])
df.isna().sum()
df = dropnan('country')
```

```python
df['customer_type'].unique()
df['customer_type'] = df['customer_type'].replace('G  ROUp', 'Group')
```

```python
df['adults'].unique()
df = df[df['adults'] != 3500 & 0 ]
df['adults'].unique()
```

```python
df['arrival_date_year'].unique()
(df['arrival_date_year']== 2099).sum()
df= df[df['arrival_date_year'] != 2099]
```

```python
df['meal'].unique()
df['meal'] = df['meal'].str.strip()
df['meal'].unique()
```

```python
df[df['booking_id'].duplicated(keep=False)]
df = df.drop_duplicates(subset = ['booking_id'])
df.head()
```
