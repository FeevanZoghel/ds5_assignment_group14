```python
print('hello')
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as st
import statsmodels.api as sm
from statsmodels.formula.api import ols
import math
```

```python
df = pd.read_excel('retail.xlsx')

df.head()

```

```python
#total sales for each category
salescat = df.groupby('Category')['Sales'].sum()
salescat
```

```python
#percentage taht a category contirbutes tot the total sales
perccat = {}
totalsales = df['Sales'].sum()
for value in salescat.index:
    percentagetotal = salescat[value] /totalsales * 100 
    perccat[value] = percentagetotal

```

```python
#sales for each month
salesmonth = df.groupby('Month')['Sales'].sum()
salesmonth
```

```python
# percentage that the month contribute to the total 
percmonth = {}

for month in salesmonth.index:
    percentagemonth = salesmonth[month] /totalsales * 100 
    percmonth[month] = percentagemonth
```

```python
#omzetten naar een dataframe
dfcat = pd.DataFrame(perccat.items(), columns = ['Category', 'Percentage'])

dfmonth = pd.DataFrame(percmonth.items(), columns = ['Month', 'Percentage'])
```

```python
# exporteren naar Excel 
with pd.ExcelWriter('nee.xlsx') as writer: 
    dfcat.to_excel(writer, sheet_name= 'Category', index = False)
    dfmonth.to_excel(writer, sheet_name= 'Month', index = False)
```
