```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as st
import statsmodels.api as sm
from statsmodels.formula.api import ols
import math
```

```python
df = pd.read_excel('detailedRetail.xlsx')
df.head()
```

```python
# category salesb
catsales = df.groupby('Category')['Sales'].sum()
print(catsales)
```

```python
# perc cat 
totalsales = df.Sales.sum()
perccat = {}

for category in catsales.index:
    percentagecat = catsales[category] /totalsales * 100
    perccat[category] = percentagecat
```

```python
# sales month
salmonth = df.groupby('Month')['Sales'].sum()
```

```python
# perc month 

percmonth = {}

for month in salmonth.index: 
    percentagemonth = salmonth[month] / totalsales * 100
    percmonth[month] = percentagemonth
```

```python
# sales man 
salesman = df.groupby('Sales Manager')['Sales'].sum()
```

```python
# perc man 
percman = {}
for man in salesman.index: 
    percentageman  = salesman[man] / totalsales * 100
    percman[man] = percentageman
```

```python
dfcat = pd.DataFrame(perccat.items(), columns = ['cat','percentage'])
dfmonth = pd.DataFrame(percmonth.items(), columns = ['month','percentage'])
dfman = pd.DataFrame(percman.items(), columns = ['manager','percentage'])
```

```python
with pd.ExcelWriter('ditiseenexcelbestand.xlsx') as writer: 
    dfcat.to_excel(writer, sheet_name= 'category', index = False)
    dfmonth.to_excel(writer, sheet_name= 'month', index = False)
    dfman.to_excel(writer, sheet_name= 'manager', index = False)
```
