```python
from langdetect import detect, LangDetectException
```

```python
df = pd.read_excel('tweets.xlsx')

df.head()
```

```python
def detect_language(text)->str| None:
    '''
    input: kolom met text

    args: 
        kijkt per regel van de kolom in welke taal het is geschreven

    return: 
        de gedetecteerde taal of None als het nisk kan detecteren
    '''
    try:
        return detect(text)
    except (LangDetectException, TypeError):
        return None

df['language'] = df['Tweet'].apply(detect_language)
```
