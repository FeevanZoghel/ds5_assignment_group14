```python
import pandas as pd
```

```python
df = pd.read_excel('hotelBookings.xlsx')
df.head()
```

```python
print(df.columns)
df = df.drop_duplicates()
```

```python
#Is al schoon

#hotel
print(df['hotel'].unique())
#is_canceled
print(df['is_canceled'].unique())
#lead_time
print(df['lead_time'].unique())
#arrival date_week_number
print(df['arrival_date_week_number'].unique())
#arrival date day of moth
print(df['arrival_date_day_of_month'].unique())
#stays in weekend nights
print(df['stays_in_weekend_nights'].unique())
#babies
print(df['babies'].unique())
#distribution channel
print(df['distribution_channel'].unique())
#is repeated quest
print(df['is_repeated_guest'].unique())
#previous cancellations
print(df['previous_cancellations'].unique())
#previous bookings not canceled
print(df['previous_bookings_not_canceled'].unique())
#reserved room type
print(df['reserved_room_type'].unique())
#assigned room type
print(df['assigned_room_type'].unique())
#booking changes
print(df['booking_changes'].unique())
#deposit_type
print(df['deposit_type'].unique())
#days in waiting list
print(df['days_in_waiting_list'].unique())
#adr
print(df['adr'].unique())
#required car parking spaces
print(df['required_car_parking_spaces'].unique())
#total of special requests
print(df['total_of_special_requests'].unique())
#reservation status
print(df['reservation_status'].unique())
#reservations status date
print(df['reservation_status_date'].unique())
```

```python
#arrival_date_year
print(df['arrival_date_year'].unique())
vijftien = 0
negen = 0
for i in df['arrival_date_year']:
    if i == 2015:
        vijftien+=1
    elif i == 2099:
        negen+=1

print(f'2015: {vijftien}')
print(f'2099: {negen}')

#cleanen
df.loc[df['arrival_date_year']==2099, 'arrival_date_year'] = 2015
df.copy()
print(df['arrival_date_year'].unique())
```

```python
#arrival date month
df = df.dropna(subset=['arrival_date_month'])
print(df['arrival_date_month'].unique())
```

```python
#stays in week nights
df['stays_in_week_nights'] = df['stays_in_week_nights'].astype(int)
print(df['stays_in_week_nights'].unique())
```

```python
#adults
x = 0

for i in df['adults']:
    if i == 3500:
        x+=1

df = df[df['adults']!= 3500]
df.copy()
print(df['adults'].unique())
```

```python
#children
print(df['children'].unique())

x = 0
y = 0
z = 0

for i in df['children']:
    if i == 200:
        x += 1
    elif i == 55000:
        y += 1
    elif i == 10:
        z+=1

print(x,y,z)

df = df[(df['children'] != 200) & (df['children'] != 55000) & (df['children'] != 10)]
df.copy()
print(df['children'].unique())
```

```python
#meal
df['meal'] = df['meal'].str.strip()
print(df['meal'].unique())
```

```python
#country
df = df[(df['country'] != 2) & (df['country'] != 3)]
df['country'] = df['country'].str.strip()
df = df.dropna(subset = 'country')
df.copy()
print(df['country'].unique())
```

```python
#market segment
df['market_segment'] = df['market_segment'].str.capitalize()
df['market_segment'] = df['market_segment'].str.replace('G  roups', 'Groups')
print(df['market_segment'].unique())
```

```python
#agent
df['agent'] = df['agent'].astype('Int64')
print(df['agent'].unique())
```

```python
#company
df['company'] = df['company'].astype('Int64')
print(df['company'].unique())
```

```python
#customer type
df['customer_type'] = df['customer_type'].str.capitalize().str.replace('G  roup', 'Group')
print(df['customer_type'].unique())
```
