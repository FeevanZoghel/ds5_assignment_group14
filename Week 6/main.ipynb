```python
print('hello')
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as st
import statsmodels.api as sm
from statsmodels.formula.api import ols
import math
```

```python
def load_data(excelbestand: str) -> pd.DataFrame:
    '''
    zet een excelbestand om tot een dataframe

    return: 
        Dataframe
        
    Gemaakt door: Matthijs
    '''
    df = pd.read_excel(excelbestand)
    return df

def check_missing_values(df: pd.DataFrame)-> list:
    '''
    Kijkt voor een DataFrame of er Nans in staan

    return: 
        ene lijst met de kolomnamen het aantal nanwaardes van die kolom

    Gemaakt door: Matthijs
    '''
    kolom_with_nan = []
    for kolom in df:
        missende_waardes = False
        if df[kolom].isna().sum() > 0:
                kolom_with_nan.append((kolom,df[kolom].isna().sum()))
                missende_waardes = True
        if missende_waardes ==True:
             print('Deze dataframe bevat missende waardes')
    return kolom_with_nan

def drop_duplicates(df: pd.DataFrame)-> pd.DataFrame:
    '''
    input: 
        dataframe

    Deze functie verwijderd dubbele rijen uit de gegeven dataframe

    return: 
        een dataframe zonder dubbele rijen
    
    Gemaakt door Matthijs
    '''
    if df.duplicated().any() == True:
        print('er zijn dubbele rijen gevonden')
        df.drop_duplicates()

```

```python
# opdracht 4

df = load_data('hotelBookings.xlsx')
check_missing_values(df)
```

```python
for value in df['arrival_date_month']:
    df['arrival_date_month'] = df[df['arrival_date_month'].isna()]
        

```

```python
#gooit degene met een nan weg
df = df.dropna(subset = ['arrival_date_month'])

#veranderd het jaartal 2099 naar 2015
df.loc[df['arrival_date_year']=='2099', 'arrival_date_year'] = 2015

# dit zorgt dat alle overbodige spaties worden verwijderd
df['meal'] = df['meal'].str.strip().str[-3:]

# alles omzetten naar een integer en overbodige waardes eruit halen
df['adults'] = df['adults'].astype(int)
df.drop(df[df['adults']==3500].index, inplace = True)
df['children'] = df['children'].astype(int)
df.drop(df[df['children']>10].index, inplace = True) 
df['babies'] = df['babies'].astype(int)
```
