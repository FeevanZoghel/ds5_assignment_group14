```python
import pandas as pd
```

```python
df = pd.read_excel('retail.xlsx')
df.head(100)
```

```python
#Mijn manier opdracht 2a
stores = {}

for index, row in df.iterrows():
    category = row['Category']
    sales = row['Sales']
    if category in stores:
        stores[category] += sales
    else:
        stores[category] = sales

print(stores)
```

```python
#die van chat
stores = df.groupby('Category')['Sales'].sum()
stores
```

```python
# 2b
total_sales = df['Sales'].sum()
sales_perc = {}

for category in stores.index:
    perc = (stores[category] / total_sales) * 100
    print(f'{category:<50} {perc:>6.2f}%')
    sales_perc[category] = perc
```

```python
#2c
sales = df.groupby('Month')['Sales'].sum()
print(sales)
```

```python
#2d
total_sales = df['Sales'].sum()
month_perc = {}

for month in sales.index:
    perc = (sales[month] / total_sales) * 100
    print(f'{month:<15} {perc:>3.2f}%')
    month_perc[month] = perc

```

```python
#3,4
with pd.ExcelWriter('reportRetail_fee.xlsx') as writer:
    pd.DataFrame(list(stores.items()), columns=['Category', 'Sales']).to_excel(
        writer, sheet_name='Sale category', index=False)

    pd.DataFrame(list(sales_perc.items()), columns=['Category', 'Percentage to total']).to_excel(
        writer, sheet_name='Category percentage', index=False)

    pd.DataFrame(list(sales.items()), columns=['Month', 'Sales']).to_excel(
        writer, sheet_name='Sale Month', index=False)

    pd.DataFrame(list(month_perc.items()), columns=['Month', 'Percentage to total']).to_excel(
        writer, sheet_name='Month percentage', index=False)
```

```python
#5
def get_total_sales(df, category = 'Category', sales= 'Sales'):
    '''
    Get the total sales per category or month

    arg:
    df:         Dataframe
    category:   category in this case either 'month' or 'category'. Standard = 'Category'
    sales:      the sales per category. Standard = 'Sales'

    return:
    a pandas frame dinge with the categories (or month) and the corresponding sales grouped by that category
    '''
    answer = df.groupby(category)[sales].sum()
    data = answer.reset_index()
    return data

def get_percentage_sales(df, category='Category', sales='Sales'):
    '''
    Get the percentage of total sales per category or month.

    Args:
        df: DataFrame containing the data.
        category: Column to group by. Default = 'Category'.
        sales: Column containing sales. Default = 'Sales'.

    Returns:
        DataFrame with the percentage of sales per category or month.
    '''
    total_sales = df[sales].sum()
    stores = df.groupby(category)[sales].sum()

    sales_perc = {}

    for i in stores.index:
        perc = (stores[i] / total_sales) * 100
        sales_perc[i] = perc

    data = pd.DataFrame(list(sales_perc.items()), columns=[category, 'Percentage'])

    return data

def export_to_excel(df, filename='reportRetail_fee.xlsx'):
    '''
    Export sales data to an Excel file with multiple sheets.

    Args:
        df: DataFrame containing the sales data.
        filename: Name of the Excel file.

    Returns:
        None
    '''
    with pd.ExcelWriter(filename) as writer:
        get_total_sales(df).to_excel(
            writer, sheet_name='Sale category', index=False
        )

        get_percentage_sales(df).to_excel(
            writer, sheet_name='Category percentage', index=False
        )

        get_total_sales(df, 'Month').to_excel(
            writer, sheet_name='Sale Month', index=False
        )

        get_percentage_sales(df, 'Month').to_excel(
            writer, sheet_name='Month percentage', index=False
        )
export_to_excel(df)
```
