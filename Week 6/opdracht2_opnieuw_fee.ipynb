```python
import pandas as pd
```

```python
df = pd.read_excel('detailedRetail.xlsx')
df.head()
```

```python
sales_cat = df.groupby('Category')['Sales'].sum()
# print(sales_cat)
total_sales = df['Sales'].sum()

sales_perc = {}
for category in sales_cat.index:
    perc = (sales_cat[category]/total_sales)*100
    sales_perc[category] = perc
category_nieuw = pd.DataFrame(sales_perc.items(), columns = ['Category', 'Sales'])
category_nieuw.head()
```

```python
sales_month = df.groupby('Month')['Sales'].sum()

month_perc = {}
for category in sales_month.index:
    perc = (sales_month[category]/total_sales)*100
    month_perc[category] = perc
month = pd.DataFrame(month_perc.items(), columns = ['Month', 'Sales'])
month.head()
```

```python
sales_manager = df.groupby('Sales Manager')['Sales'].sum()

manager_perc = {}
for category in sales_manager.index:
    perc = (sales_manager[category]/total_sales)*100
    manager_perc[category] = perc  
manager = pd.DataFrame(manager_perc.items(), columns = ['Manager', 'Sales'])
manager.head()
```

```python
with pd.ExcelWriter('naam_fee.xlsx') as writer:
    category_nieuw.to_excel(sheet_name = 'Category', index = False)
    month.to_excel(sheet_name = 'month', index = False)
    manager.to_excel(sheet_name = 'manager', index = False)
```
