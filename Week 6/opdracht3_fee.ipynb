```python
import pandas as pd
from langdetect import detect, LangDetectException
from textblob import TextBlob
```

```python
df = pd.read_excel('tweets.xlsx')
df.head()
```

```python
def detect_language(tweet):
    try:
        return detect(str(tweet))
    except LangDetectException:
        return 'Unknown'

language = {}

for i in df['Tweet']:
    language[i] = detect_language(i)
```

```python
df2 = pd.DataFrame(list(language.items()), columns = ['Tweet', 'Language'])
df2.head()
```

```python
text = "TextBlob is amazingly simple to use. It makes NLP tasks fun!"
blob = TextBlob(text)

print(blob.sentiment)
# Sentiment(polarity=0.625, subjectivity=0.6)

print(blob.sentiment.polarity)    # 0.625
print(blob.sentiment.subjectivity) # 0.6
```
