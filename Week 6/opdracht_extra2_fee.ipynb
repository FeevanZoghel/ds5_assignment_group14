```python
import pandas as pd
```

```python
df = pd.read_excel('DS5_oefendataset_data_cleaning_handling.xlsx', sheet_name = 'HotelBookings')
df = df.copy()
df = df.drop_duplicates()
df.head()
```

```python
df['adults'].unique()
```

```python
def onnodige_waarden(df, column_name = 'arrival_date_year', value = '2099'):
    '''
    Haalt de onnodige waarden uit een kolom uit een dataframe
    arg:
        df = dataframe
        column_name = kolomnaam in dataframe. Normaal: 'arrival_date_year'
        value = de waardie die onnodig is dus eruit wordt gehaald. Normaal: '2099'
    return: 
        niks, omdat die al print
    '''

    df = df.copy()
    df_nieuw = df[(df[column_name]!= value)]
    print(df_nieuw[column_name].unique())
    df = df_nieuw

onnodige_waarden(df,'arrival_date_year', '2099')
```

```python
df['arrival_date_month'] = df['arrival_date_month'].str.capitalize()
df = df.dropna(subset = 'arrival_date_month')
df['arrival_date_month'].unique()
```

```python
def type_veranderen(df, column_name = 'adults', value = 'Int64'):
    '''
    Verandert het type van de waardes in de kolom
    Meestal nodig om een float naar een integer waarde te veranderen
    arg:
        df = dataframe
        column_name = kolomnaam in de dataframe. Standaard: 'adults'
        value = Het type waarin je de waardes wilt veranderen. Standaard: 'Int64' (integer)

    return:
        Niks want hij print al
    '''

    df_nieuw = df.copy()
    df_nieuw[column_name] = df[column_name].astype(value)
    df = df_nieuw
    print(df[column_name].unique())

type_veranderen(df,'adults', 'Int64')
onnodige_waarden(df,'adults', '3500')
df = df.dropna(subset = 'adults')
df['adults'].unique()
```
