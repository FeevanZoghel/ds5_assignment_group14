```python
import pandas as pd
```

```python
df = pd.read_excel('oefendataset_hotelboekingen.xlsx')
df = df.drop_duplicates()
df.head()
```

```python
df['meal'].unique()
```

```python
df = df.dropna(subset = 'country')
df['country'].unique()
```

```python
df['customer_type'] = df['customer_type'].str.replace('G  ROUp', 'Group')
df['customer_type'].unique()
```

```python
df = df[(df['adults']!= 3500)]
df['adults'].unique()
```

```python
df['price_per_night'] = df['price_per_night'].astype(float)
df['price_per_night'].unique()
```

```python
df = df[(df['arrival_date_year']!= 2099)]
df['arrival_date_year'].unique()
```

```python
df['meal'] = df['meal'].str.replace(' No Meal ', 'No Meal')
df['meal'].unique()
```
