```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as st
import statsmodels.api as sm
from statsmodels.formula.api import ols
import math
```

```python
df = pd.read_excel('hotelBookings.xlsx')
df.head()
```

```python
for kolom in df:
    print(f'{df[kolom].unique()} --{kolom}--')
```

```python
#unieke waarde maken

df.loc[df['arrival_date_year']== 2099, 'arrival_date_year'] = 2015
```

```python

df = df.dropna(subset=['arrival_date_month'])
df = df.dropna(subset = ['country'])

for kolom in df:
    print(f'{kolom} -----{df[kolom].isna().sum()}')
  
```

```python
df['adults'].unique()
df.loc[df['adults']==3500, 'adults'].count()
df = df.drop(df[df['adults']==3500].index)
```

```python
df['children'].unique()
df = df.drop(df[df['children']==55000].index)
```

```python
df['stays_in_week_nights'].unique()
df['stays_in_week_nights'] = df['stays_in_week_nights'].astype(int)
df['stays_in_week_nights'].unique()
```

```python
df['meal'] = df['meal'].str.strip()
```

```python

df = df[df['country'] != 3]
df['country'] = df['country'].str.strip()
df.unique()
```

```python

df['market_segment'] = df['market_segment'].str.strip()
df.loc[df['market_segment'] == 'G  ROUps', 'market_segment'] ='Groups'

df['market_segment'].unique()
```

```python
df.drop_duplicates()
```

```python
# company
df['company'] = df['company'].astype('Int64')
df['company'].unique()
```

```python
df['customer_type'].unique()
df['customer_type'] = df['customer_type'].str.replace('G  ROUp', 'Group')
df['customer_type'].unique()
```
